```python
def is_even(num):
    if  type(num)==str:
        return "Entered number not be string"

    else:    
        if num%2==0:
            print('even')
        else:
            print('odd')

is_even(2)
```

```python
def multiply(*mul):
    prod =1
    for i in mul:
        prod = prod+i
    return prod
```

```python
def display(**kwargs):
    for (key,value) in kwargs.items():
        print(key,'---',value)
display(pak="isdb", ind='delhi')
```

```python
def place_order(customer_name, *items, **details):
    print (f"Customer Name : {customer_name} \n")
    if len(items)==0:
        print('No items selected')
    else:
        print("Ordered Items")
        for item in items:
            print(item)
    print('Customer Detals \n')
    for (key,value) in details.items():
        print(f'{key} : {value}')
    if 'phone' not in details:
        print('phone number is missing \n')

    print(f"\n Number of Items : {len(items)}")
place_order("Umar",'pizza','burger','mango',city='Lahor',id='123')
```
